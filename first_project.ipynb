```python
import numpy as np

np.random.seed(42)

data = np.array([
    [2.0, 60, 55],
    [3.5, 75, 68],
    [5.0, 80, 78],
    [1.5, 50, 45],
    [6.0, 90, 88],
    [4.0, 85, 72],
    [7.0, 95, 93],
    [2.5, 65, 60],
    [5.5, 88, 84],
    [3.0, 70, 65]
], dtype=np.float64)

# Columns: study hours, attendance %, exam score

print("Dataset:\n", data)
print("\nShape:", data.shape)
print("Dimensions:", data.ndim)
print("Total elements:", data.size)
print("Data type:", data.dtype)
```

```python
study_hours = data[:, 0]
attendance = data[:, 1]
exam_scores = data[:, 2]

print("Average exam score:", np.mean(exam_scores))
print("Highest exam score:", np.max(exam_scores))
print("Lowest exam score:", np.min(exam_scores))
print("Score standard deviation:", np.std(exam_scores))

# Find students who scored at least 75
high_scorers = data[data[:, 2] >= 75]

print("\nStudents scoring at least 75:\n", high_scorers)

# Correlation between study hours and exam scores
correlation = np.corrcoef(study_hours, exam_scores)[0, 1]
print("\nStudy hours / exam score correlation:", correlation)
```
