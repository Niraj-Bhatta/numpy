```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "age": [20, 22, np.nan, 25, 22],
    "hours_studied": [2, 5, 3, 8, 5],
    "score": [55, 75, 65, 95, 75],
    "passed": [False, True, True, True, True]
})
```

```python
df
```

```python
print(df.head())
print(df.shape)
print(df.dtypes)
print(df.describe())

# Select a column
print(df["score"])

# Select rows where the score is at least 70
print(df[df["score"] >= 70])

# Select specific rows and columns
print(df.loc[0:2, ["age", "score"]])
```

```python
# Count missing values per column
print(df.isna().sum())

# Fill missing ages with the median
df["age"] = df["age"].fillna(df["age"].median())

# Remove duplicate records
df = df.drop_duplicates()

print(df.isna().sum())
```

#Fearure engineering


```python
# Create a new feature
df["study_efficiency"] = (
    df["score"] / df["hours_studied"]
)

# Categorize scores
df["performance"] = np.where(
    df["score"] >= 70,
    "Good",
    "Needs improvement"
)

# Summarize the data
print(df.groupby("passed")["score"].mean())
```
